```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

def generate_gif(th1, th2, filename="animation.gif"):
    # Angles from 0° to 180° in steps of 15°
    angles = np.arange(th1, th2, 15)

    # Generate x values for plotting the sine curve
    x = np.linspace(th1, th2, 500)
    y = np.sin(np.radians(x))

    fig, ax = plt.subplots(figsize=(8, 5))

    ax.set_xlim(th1, th2)
    ax.set_ylim(-2, 2)
    ax.set_xlabel("Angle (degrees)")
    ax.set_ylabel("sin(x)")
    ax.set_title("Sine Function Animation")
    ax.grid(True)

    # Plot the complete sine curve
    ax.plot(x, y, color="lightgray", label="sin(x)")

    # Animated point and vertical line
    point, = ax.plot([], [], "ro", markersize=8)
    line, = ax.plot([], [], "r--", linewidth=1.5)
    label = ax.text(0.05, 0.90, "", transform=ax.transAxes, fontsize=12)

    def update(frame):
        angle = angles[frame]
        value = np.sin(np.radians(angle))

        point.set_data([angle], [value])
        line.set_data([angle, angle], [0, value])
        label.set_text(
            f"x = {angle}°\nsin(x) = {value:.3f}"
        )

        return point, line, label

    animation = FuncAnimation(
        fig,
        update,
        frames=len(angles),
        interval=500,
        blit=True,
        repeat=True
    )

    animation.save(filename, writer=PillowWriter(fps=2))
    plt.close(fig)

    print(f"GIF saved as {filename}")



generate_gif(-180, 180)
```
